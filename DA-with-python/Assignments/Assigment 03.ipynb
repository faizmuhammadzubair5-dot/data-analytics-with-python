 3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard.

```python
numbers = [int(value) for value in input("Enter numbers separated by space: ").split()]

if numbers:
    largest = numbers[0]

    for number in numbers[1:]:
        if number > largest:
            largest = number

    print("Largest number:", largest)
else:
    print("The list is empty.")
```

 4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index

```python
items = input("Enter list elements separated by space: ").split()

if items:
    rotated_items = items[-1:] + items[:-1]
    print("Rotated list:", rotated_items)
else:
    print("The list is empty.")
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It
should print the date in the form March 12, 2021. 


```python
from datetime import datetime

date_input = input("Enter a date (mm/dd/yyyy): ")

date = datetime.strptime(date_input, "%m/%d/%Y")

print(f"Formatted date: {date.strftime('%B')} {date.day}, {date.year}")
```

 7. Write a program with a function that accepts a string from keyboard and create a new string after
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses."
the output should be "Stop And Smell The Roses"

```python
def capitalize_words():
    sentence = input("Enter a sentence: ")
    return " ".join(word.capitalize() for word in sentence.split())


print(capitalize_words())
```

9. Write a program to add two matrices of size n x m.

```python
rows = int(input("Enter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter elements of first matrix:")
for i in range(rows):
    row = list(map(int, input(f"Row {i + 1}: ").split()))
    matrix1.append(row)

print("Enter elements of second matrix:")
for i in range(rows):
    row = list(map(int, input(f"Row {i + 1}: ").split()))
    matrix2.append(row)

sum_matrix = []

for i in range(rows):
    row = []
    for j in range(columns):
        row.append(matrix1[i][j] + matrix2[i][j])
    sum_matrix.append(row)

print("Sum of the two matrices:")
for row in sum_matrix:
    print(row)
```

10. Write a program to multiply two matrices 

```python
rows1 = int(input("Enter rows of first matrix: "))
columns1 = int(input("Enter columns of first matrix: "))

rows2 = int(input("Enter rows of second matrix: "))
columns2 = int(input("Enter columns of second matrix: "))

if columns1 != rows2:
    print("Matrix multiplication is not possible.")
else:
    matrix1 = []
    matrix2 = []

    print("Enter elements of first matrix:")
    for i in range(rows1):
        matrix1.append(list(map(int, input(f"Row {i + 1}: ").split())))

    print("Enter elements of second matrix:")
    for i in range(rows2):
        matrix2.append(list(map(int, input(f"Row {i + 1}: ").split())))

    product = []

    for i in range(rows1):
        row = []
        for j in range(columns2):
            value = 0
            for k in range(columns1):
                value += matrix1[i][k] * matrix2[k][j]
            row.append(value)
        product.append(row)

    print("Product of the two matrices:")
    for row in product:
        print(row)
```
