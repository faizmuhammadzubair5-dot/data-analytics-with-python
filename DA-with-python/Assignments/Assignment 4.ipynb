Python Assignment — Dictionaries Practice

Q1 — Basic Dictionary Creation

```python
student = {
    "name": "Faiz",
    "age": 20,
    "course": "Computer Science",
    "city": "karachi"
}

for value in student.values():
    print(value)
```

Q2 — Adding & Updating Values

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

car["color"] = "White"
car["year"] = 2022

print(car)
```

Q3 — Loop Through Dictionary


```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers


```python
numbers = [1, 2, 2, 3, 3, 3, 4]

frequency = {}

for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1

print(frequency)
```

Q5 — Word Counter

```python
sentence = input("Enter a sentence: ")
word_frequency = {}

for word in sentence.split():
    word_frequency[word] = word_frequency.get(word, 0) + 1

print(word_frequency)
```

Q6 — Student Marks & Average

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = sum(marks.values())
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

Q7 — Highest Scoring Student

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = max(students, key=students.get)

print("Highest scoring student:", highest_student)
print("Score:", students[highest_student])
```

Q8 — Merge Dictionaries

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```

Q9 — Mini Project: Contact Book

```python
contacts = {}

for _ in range(3):
    name = input("Enter contact name: ")
    phone = input("Enter phone number: ")
    contacts[name] = phone

search_name = input("Search by name: ")

if search_name in contacts:
    print("Phone:", contacts[search_name])
else:
    print("Contact not found")
```
