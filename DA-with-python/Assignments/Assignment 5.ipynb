
QUESTION 2: Dictionary Basics


1.	Create a dictionary called student with the following keys:

o	name

o	age

o	course

o	marks (must store a list of 3 marks)


```python
student = {
    "name": "Alex",
    "age": 20,
    "course": "Python",
    "marks": [85, 90, 88]
}

student
```

2.	Using loops:

o	Print all keys and their values.

o	Calculate the average of the marks.

o	Display whether the student Passed (average ≥ 50) or Failed.


```python
print("Student details:")
for key, value in student.items():
    print(key, ":", value)

marks_total = 0
for mark in student["marks"]:
    marks_total += mark

average_marks = marks_total / len(student["marks"])
print("Average marks:", average_marks)

if average_marks >= 50:
    print("Passed")
else:
    print("Failed")
```

Question 3 : List of Dictionaries

1 : Create a list calles employees

```python
employees = [
    {"name": "Sam", "age": 28, "department": "HR"},
    {"name": "Jordan", "age": 35, "department": "IT"},
    {"name": "Taylor", "age": 31, "department": "Sales"}
]
```

2 : Each employee must be stored as a dictionary containing

name 

department

salary

```python
for employee, salary in zip(employees, [50000, 65000, 58000]):
	employee["salary"] = salary

employees
```

3 : Using loops

print details of all  employees

Find the employee with the highest salary

Calculate the total salary expense

```python
highest_paid = employees[0]
total_salary = 0

for employee in employees:
    if employee["salary"] > highest_paid["salary"]:
        highest_paid = employee
    total_salary += employee["salary"]

print("Employee with the highest salary:", highest_paid["name"])
print("Salary:", highest_paid["salary"])
print("Total salary expense:", total_salary)
```

2 : Store the numbers in a list

```python
entered_numbers = []

while True:
    user_input = input("Enter a number (or press Enter to finish): ")

    if user_input == "":
        break

    try:
        entered_numbers.append(int(user_input))
    except ValueError:
        print("Invalid input. Please enter a valid integer.")

print("Numbers entered:", entered_numbers)
```

3 : Stop when the user enters -1

```python
while True:
    user_input = input("Enter a number (-1 to stop): ")

    try:
        number = int(user_input)
    except ValueError:
        print("Invalid input. Please enter an integer.")
        continue

    if number == -1:
        break

    entered_numbers.append(number)

print("Numbers entered:", entered_numbers)
```

4 : After the loop ends: 

Print the complete list

Print the largest number

Print the smallest number

```python
print("Complete list:", entered_numbers)

if entered_numbers:
    print("Largest number:", max(entered_numbers))
    print("Smallest number:", min(entered_numbers))
else:
    print("No numbers were entered.")
    
```

Question 5 : Word Frequency Counter

1.	Ask the user to enter a sentence.

```python
sentence = input("Enter a sentence: ")

print("You entered:", sentence)

```

2.	Convert the sentence into a list of words

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

print("You entered:", sentence)
print("Words:", words)
```

3.	Use a dictionary to count how many times each word appears.

```python
word_counts = {}

for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print("Word frequencies:", word_counts)
```

4.	Print the word frequency clearly.



```python
for word, count in word_counts.items():
    print(f"{word}: {count}")
```

QUESTION 6: Nested Loops – Multiplication Table

```python
# Print a multiplication table from 1 to 5 using nested loops.
for row in range(1, 6):
    for column in range(1, 6):
        print(f"{row * column:2}", end=" ")
    print()  # Start a new row.
```
